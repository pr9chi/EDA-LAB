```python
bins = [0, 50, 100, 500, 1000, np.inf]

bin_names = [
    'Low',
    'Medium',
    'High',
    'Very High',
    'Extreme'
]

Fraud_train['amt_bin'] = pd.cut(
    Fraud_train['amt'],
    bins=bins,
    labels=bin_names
)

Fraud_train[['amt', 'amt_bin']].head(10)
```

```python
pd.value_counts(Fraud_train['amt_bin'])
```

```python
Fraud_test['amt_bin'] = pd.cut(
    Fraud_test['amt'],
    bins=bins,
    labels=bin_names
)

Fraud_test[['amt', 'amt_bin']].head(10)
```

```python
Fraud_train['amt_quantile_bin'] = pd.qcut(
    Fraud_train['amt'],
    4,
    labels=['Q1', 'Q2', 'Q3', 'Q4']
)

Fraud_train[['amt', 'amt_quantile_bin']].head(10)
```

```python
pd.value_counts(Fraud_train['amt_quantile_bin'])
```

```python
numeric_features = [
    'amt',
    'lat',
    'long',
    'city_pop',
    'merch_lat',
    'merch_long'
]

train_numeric = Fraud_train[numeric_features]
test_numeric = Fraud_test[numeric_features]
```

```python
min_max_scaler = MinMaxScaler()

Fraud_train_min_max = pd.DataFrame(
    min_max_scaler.fit_transform(train_numeric),
    columns=numeric_features
)

Fraud_train_min_max.head()
```

```python
Fraud_test_min_max = pd.DataFrame(
    min_max_scaler.transform(test_numeric),
    columns=numeric_features
)

Fraud_test_min_max.head()
```

```python
standard_scaler = StandardScaler()

Fraud_train_standard = pd.DataFrame(
    standard_scaler.fit_transform(train_numeric),
    columns=numeric_features
)

Fraud_train_standard.head()
```

```python
Fraud_test_standard = pd.DataFrame(
    standard_scaler.transform(test_numeric),
    columns=numeric_features
)

Fraud_test_standard.head()
```

```python
robust_scaler = RobustScaler()

Fraud_train_robust = pd.DataFrame(
    robust_scaler.fit_transform(train_numeric),
    columns=numeric_features
)

Fraud_train_robust.head()
```

```python
Fraud_test_robust = pd.DataFrame(
    robust_scaler.transform(test_numeric),
    columns=numeric_features
)

Fraud_test_robust.head()
```

```python
Fraud_train['amt_zscore'] = zscore(Fraud_train['amt'])

Fraud_train[['amt', 'amt_zscore']].head(10)
```

```python
zscore_outliers = Fraud_train[
    np.abs(Fraud_train['amt_zscore']) > 3
]

zscore_outliers.head()
```

```python
print("Number of outliers:", zscore_outliers.shape[0])
```

```python
zscore_outliers['is_fraud'].value_counts()
```

```python
fraud_percentage = zscore_outliers['is_fraud'].mean() * 100

print("Fraud percentage among detected outliers:",
      fraud_percentage, "%")
```

```python
Q1 = Fraud_train['amt'].quantile(0.25)
Q3 = Fraud_train['amt'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
iqr_outliers = Fraud_train[
    (Fraud_train['amt'] < lower_bound) |
    (Fraud_train['amt'] > upper_bound)
]

iqr_outliers.head()
```

```python
print("Number of IQR outliers:", iqr_outliers.shape[0])
```

```python
iqr_outliers['is_fraud'].value_counts()
```

```python
fraud_percentage_iqr = iqr_outliers['is_fraud'].mean() * 100

print("Fraud percentage among IQR outliers:",
      fraud_percentage_iqr, "%")
```

```python
test_iqr_outliers = Fraud_test[
    (Fraud_test['amt'] < lower_bound) |
    (Fraud_test['amt'] > upper_bound)
]

test_iqr_outliers.head()
```

```python
print("Number of test outliers:", test_iqr_outliers.shape[0])
```

```python
print("Actual labels of detected test outliers:")
print(test_iqr_outliers['is_fraud'].value_counts())
```

```python
test_fraud_percentage = (
    test_iqr_outliers['is_fraud'].mean() * 100
)

print(
    "Percentage of detected test outliers that are fraudulent:",
    test_fraud_percentage,
    "%"
)
```

```python
comparison = pd.DataFrame({
    'Actual Fraud': [
        Fraud_test['is_fraud'].sum(),
        Fraud_test['is_fraud'].value_counts().get(0, 0)
    ],
    'Detected IQR Outliers': [
        test_iqr_outliers['is_fraud'].sum(),
        (test_iqr_outliers['is_fraud'] == 0).sum()
    ]
}, index=['Fraud', 'Legitimate'])

comparison
```

```python
fraudulent_outliers = test_iqr_outliers[
    test_iqr_outliers['is_fraud'] == 1
]

fraudulent_outliers[
    ['trans_date_trans_time', 'cc_num', 'merchant',
     'category', 'amt', 'city', 'state', 'is_fraud']
].head(20)
```
