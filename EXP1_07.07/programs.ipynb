```python
num = int(input("Enter a number: "))
f = 1

for i in range(1, num + 1):
    f*= i

print("Factorial:", f)
```

```python
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
```

```python
text = input("Enter a string: ")
count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print("Vowel count:", count)
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Operations:")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
op = int(input("Enter operator number: "))

if op == 1:
    print(a + b)
elif op == 2:
    print(a - b)
elif op == 3:
    print(a * b)
elif op == 4:
    if b != 0:
        print(a / b)
    else:
        print("Division by zero is not allowed")
else:
    print("Invalid operator")
```

```python
num = int(input("Enter a number: "))
count = 0

while num > 0:
    num //= 10
    count += 1

print("Number of digits:", count)
```

```python
n = int(input("Enter n: "))
sum = 0

for i in range(1, n + 1):
    sum += i

print("Sum =", sum)
```

```python
n = int(input("Number of elements: "))

numbers = []

for i in range(n):
    numbers.append(int(input("Enter number: ")))

smallest = min(numbers)
print("Smallest number:", smallest)

largest = max(numbers)
print("Largest number:", largest)
```

```python
num = int(input("Enter a number: "))

digits = len(str(num))
temp = num
total = 0

while temp > 0:
    digit = temp % 10
    total += digit ** digits
    temp //= 10

if total == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```
