```python
import pandas as pd
import numpy as np
```

```python
df1SE =  pd.DataFrame({ 'StudentID': [1,3,5,7,9],
                       'ScoreSE' : [34,11,55,82,90]})

df2SE =  pd.DataFrame({'StudentID': [2,4,6,8,10],
                       'ScoreSE': [50,22,56,18,99]})

df1ML =  pd.DataFrame({ 'StudentID': [1,3,5,7,9],
                       'ScoreML' : [70,99,17,58,89]})

df2ML =  pd.DataFrame({'StudentID': [2, 4, 6, 8, 10],
                       'ScoreML': [93, 44, 78, 97, 87]})
```

```python
#Consolidate the Course Data

dfSE = pd.concat([df1SE, df2SE], ignore_index=True)
dfML = pd.concat([df1ML, df2ML], ignore_index=True)

df = pd.concat([dfML, dfSE], axis=1)
df
```

```python
# Calculate Totals with an Outer Join

df_total = dfSE.merge(dfML, on='StudentID', how='outer')
total_count = len(df_total)
total_count
```

```python
# Outer Join

dfSE = pd.concat([df1SE, df2SE], ignore_index=True)
dfML = pd.concat([df1ML, df2ML], ignore_index=True)

df = dfSE.merge(dfML, how='outer')
df
```

```python
# Left Join

dfSE = pd.concat([df1SE, df2SE], ignore_index=True)
dfML = pd.concat([df1ML, df2ML], ignore_index=True)

df = dfSE.merge(dfML, how='left')
df
```

```python
# Right Join

dfSE = pd.concat([df1SE, df2SE], ignore_index=True)
dfML = pd.concat([df1ML, df2ML], ignore_index=True)

df = dfSE.merge(dfML, how='right')
df
```

**Reshaping with Hierarchical Indexing**

```python
data = np.arange(15).reshape((3,5))
indexers = ['Rainfall', 'Humidity', 'Wind']
dframe1 = pd.DataFrame(data, index=indexers, columns=['Delhi', 'Bangalore', 'Mumbai', 'Chennai', 'Panaji'])
dframe1
```

```python
stacked = dframe1.stack()
stacked
```

```python
stacked.unstack()
```

```python
series1 = pd.Series([000, 111, 222, 333], index=['zeros','ones', 'twos', 'threes'])
series2 = pd.Series([444, 555, 666], index=['fours', 'fives', 'sixs'])

frame2 = pd.concat([series1, series2], keys=['Number1', 'Number2'])
frame2.unstack()
```
