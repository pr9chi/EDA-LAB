```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("WineQT.csv")

df.head()
```

```python
print("First 3 rows:")
print(df.head(3))

print("\nLast 3 rows:")
print(df.tail(3))
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
mean = df["alcohol"].mean()
median = df["alcohol"].median()
mode = df["alcohol"].mode()

print(mean, median, mode)
```

```python
df.quality.value_counts().nlargest(30).plot(kind='bar', figsize=(10,6))
plt.title("Number of wines by quality")
plt.ylabel("Number of wines")
plt.xlabel("Quality")
plt.show()
```

```python
sns.FacetGrid(df).map(sns.distplot, "alcohol").add_legend()
plt.show()
```

```python
sns.FacetGrid(df).map(sns.distplot, "quality").add_legend()
plt.show()
```

```python
sns.boxplot(x="alcohol", data=df)
plt.show()
```

```python
plt.scatter(df["alcohol"], df["quality"])
plt.title("Scatter Plot for alcohol vs quality")
plt.xlabel("alcohol")
plt.ylabel("quality")
plt.show()
```

```python
sns.boxplot(x="quality", y="alcohol", data=df)
plt.show()
```

```python
sns.boxplot(x="quality", y="volatile acidity", data=df)
plt.show()
```

```python
sns.pairplot(
    df,
    vars=["fixed acidity", "volatile acidity", "alcohol"],
    kind="reg"
)
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)

sns.pairplot(
    df,
    height=2,
    vars=["fixed acidity", "volatile acidity", "alcohol"],
    hue="quality"
)

plt.show()
```

```python
from scipy import stats

corr = stats.pearsonr(df["alcohol"], df["quality"])

print("p-value:\t", corr[1])
print("cor:\t\t", corr[0])
```

```python
correlation = df.corr(method="pearson", numeric_only=True)

correlation
```

```python
sns.heatmap(
    correlation,
    xticklabels=correlation.columns,
    yticklabels=correlation.columns
)

plt.show()
```

```python
sns.heatmap(
    correlation,
    xticklabels=correlation.columns,
    yticklabels=correlation.columns,
    annot=True
)

plt.show()
```
